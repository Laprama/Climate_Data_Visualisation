## Notebook Info

In this notebook we plot color bars. <br>
This is useful for interactive 3D plots in Unity as the color bar can be placed next to a 3D Earth.

## Code

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.colorbar as mcolorbar
```

```python
# Create a figure and axes for the color bar
fig, ax = plt.subplots(figsize=(1, 6) , dpi = 200 )
fig.subplots_adjust(left=0.5)


# Define the colormap and normalization
cmap = 'magma'

vmin = 1750 
vmax = 1950

norm = mcolors.Normalize(vmin=vmin, vmax=vmax)

# Create the color bar
cb = mcolorbar.ColorbarBase(ax, cmap=cmap, norm=norm, orientation='vertical')
cb.set_label('Methane (ppb)', fontsize=20) # You can change 14 to any size you like

cb.ax.tick_params(labelsize=18) # You can change 12 to any size you like



# Save the figure
plt.savefig("methane_colorbar_transparent.png", bbox_inches='tight', transparent = True,  dpi = 300)
```

```python
# Create a figure and axes for the color bar
fig, ax = plt.subplots(figsize=(1, 6) , dpi = 200 )
fig.subplots_adjust(left=0.5)


# Define the colormap and normalization
cmap = 'viridis_r'
vmin = 1
vmax = 10
norm = mcolors.Normalize(vmin=vmin, vmax=vmax)

# Create the color bar
cb = mcolorbar.ColorbarBase(ax, cmap=cmap, norm=norm, orientation='vertical')
cb.set_label('Count', fontsize=20) # You can change 14 to any size you like

cb.ax.tick_params(labelsize=18) # You can change 12 to any size you like



# Save the figure
plt.savefig("density_colorbar_white_background.png", bbox_inches='tight', transparent = False,  dpi = 300)
```
